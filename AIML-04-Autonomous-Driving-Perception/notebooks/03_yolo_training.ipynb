# 03 - YOLOv8 training on BDD100K
Uses `configs/data.yaml`. Save `best.pt` back to Google Drive when done.

```python
from ultralytics import YOLO
# Optional experiment tracking:
# import wandb; wandb.login()
# from ultralytics import settings; settings.update({'wandb': True})
```

```python
model = YOLO('yolov8n.pt')   # change variant as needed
model.train(data='configs/data.yaml', epochs=50, imgsz=640, batch=16, project='runs/detect', name='train')
```

```python
metrics = model.val()
print('mAP50:', metrics.box.map50, '| mAP50-95:', metrics.box.map)
```

```python
import pandas as pd, matplotlib.pyplot as plt
res = pd.read_csv('runs/detect/train/results.csv'); res.columns = res.columns.str.strip()
res[['metrics/mAP50(B)', 'metrics/mAP50-95(B)']].plot(figsize=(8, 3), title='Validation mAP per epoch')
plt.show()
```
